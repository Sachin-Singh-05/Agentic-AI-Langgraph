```python
from langgraph.graph import StateGraph,START,END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
model = ChatGroq(
    model="llama-3.3-70b-versatile",
    api_key=os.getenv("GROQ_API_KEY")
)
```

```python
# creating class 
class BlogState(TypedDict):
    title:str
    outline:str
    blog_content:str
    evaluate:str
```

```python
def create_outline(state: BlogState) ->BlogState:
    # fetch title
    title = state['title']

    # call llm generate outline 
    prompt = f'Generate a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content

    # update state
    state['outline']= outline

    return state
```

```python
def create_blog(state : BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f'Write a detailed blog on the - {title} using the following outline \n {outline}'

    blog_content = model.invoke(prompt).content

    state['blog_content'] = blog_content

    return state
```

```python
def evaluate(state : BlogState) -> BlogState:
    blog_content = state['blog_content']

    prompt= f'Rate out of 100 marks and evaluate the following blog - {blog_content}'

    evaluate = model.invoke(prompt).content

    state['evaluate']=evaluate

    return state 
```

```python
graph = StateGraph(BlogState)

# add nodes 
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('evaluate',evaluate)

# add edges
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','evaluate')
graph.add_edge('evaluate',END)

workflow = graph.compile()
```

```python
intial_state = {'title': 'Rise of AI in India'}

final_state = workflow.invoke(intial_state)

print(final_state)
```
